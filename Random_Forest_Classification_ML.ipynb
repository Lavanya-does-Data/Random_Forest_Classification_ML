```python
import pandas as pd
file=pd.read_csv(r'C:\Users\Administrator\Downloads\BNPL_Financial_Default_Risk_Dataset.csv')
```

```python
x=file.drop(['Customer_ID','Default_Risk'],axis=1)
y=file['Default_Risk']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=67)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
c_cols=['Employment_Status','Late_Payment_History','Shopping_Category_Most_Frequent']
p=ColumnTransformer(transformers=[('picnic',OneHotEncoder(handle_unknown='ignore',sparse_output=False),c_cols)],remainder='passthrough')
```

```python
x_train=p.fit_transform(x_train)
x_test=p.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=100,random_state=67)
rf.fit(x_train,y_train)
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
accuracy=accuracy_score
print('Accuracy Score:',accuracy_score(y_test,y_pred))
print('~~~'*25)
cm=confusion_matrix(y_test,y_pred)
print('Confusion Matrix:',cm)
print('~~~'*25)
cr=classification_report(y_test,y_pred)
print('Classification Report:',cr)
```
