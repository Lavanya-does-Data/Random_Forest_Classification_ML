```python
import pandas as pd
file=pd.read_csv(r'C:\Users\Administrator\Downloads\used_cars\used_car_prices.csv')
```

```python
x=file.drop(['selling_price'],axis=1)
y=file['selling_price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=67)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
p=ColumnTransformer(transformers=[('picnic',OneHotEncoder(handle_unknown='ignore',sparse_output=False),['brand','fuel','owner'])],remainder='passthrough')
```

```python
x_train=p.fit_transform(x_train)
x_test=p.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor()
rf.fit(x_train,y_train)
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import (r2_score,mean_absolute_error,mean_squared_error)
print('R square score:',r2_score(y_test,y_pred))
print('Mean absolute error:',mean_absolute_error(y_test,y_pred))
print('Mean squared error',mean_squared_error(y_test,y_pred))
```
